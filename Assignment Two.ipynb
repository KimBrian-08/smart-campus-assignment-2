### Assignment Two: Smart Campus Service System

**Week 7 – Polymorphism**

This notebook demonstrates the use of:

- Encapsulation
- Inheritance
- Polymorphism
- Method overriding

The program models different campus services using a base class and two subclasses.

```python
class CampusService:
    """
    This creates a Base class for all campus services.
    Demonstrates encapsulation using a private attribute and getter and setter methods.
    """

    def __init__(self, service_name: str, location: str, fee: float) -> None:
        self.service_name = service_name
        self.location = location
        self.__fee = 0.0   # private attribute initiallized to zero.
        self.set_fee(fee)  # setter method for validation of fee input to be non-negative.

    def get_fee(self) -> float:
        """
        Getter method for the private fee attribute.
        """
        return self.__fee

    def set_fee(self, fee: float) -> None:
        """
        Setter method for the private fee attribute.
        Prevents a negative fee from being stored.
        """
        if fee < 0:
            raise ValueError("Fee cannot be negative.")
        self.__fee = fee

    def display_service(self) -> str:
        """
        Displays information about the campus service.
        """
        return (
            f"Service: {self.service_name}\n"
            f"Location: {self.location}\n"
            f"Fee: Kshs {self.get_fee():.2f}" # gets fee using the getter method to two decimal places.
        )
```

```python
class LibraryService(CampusService):
    """
    Subclass of CampusService representing a library services.
    """

    def __init__(self, service_name: str, location: str, fee: float, study_rooms: int, ebook_access: bool, max_borrow_limit: int) -> None:
        super().__init__(service_name, location, fee)
        self.study_rooms = study_rooms
        self.ebook_access = ebook_access
        self.max_borrow_limit = max_borrow_limit

    def display_service(self):
        """
        Overrides the base class method to include library-specific details.
        """
        base_info = super().display_service()
        ebook_status = "Available" if self.ebook_access else "Not Available"

        return (
            f"{base_info}\n"
            f"Study Rooms: {self.study_rooms}\n"
            f"E-book Access: {ebook_status}\n"
            f"Max Borrow Limit: {self.max_borrow_limit}"
        )


class ShuttleService(CampusService):
    """
    Subclass of CampusService representing a campus shuttle service.
    """

    def __init__(self, service_name: str, location: str, fee: float, route_code: str, weekend_service: bool) -> None:
        super().__init__(service_name, location, fee)
        self.route_code = route_code
        self.weekend_service = weekend_service

    def display_service(self) -> str:
        """
        Overrides the base class method to include shuttle-specific details.
        """
        base_info = super().display_service()
        weekend_status = "Available" if self.weekend_service else "Not Available"

        return (
            f"{base_info}\n"
            f"Route Code: {self.route_code}\n"
            f"Weekend Service: {weekend_status}"
        )
```

```python
# Create objects from the subclasses
library_service = LibraryService(
    service_name="AI Ogura Foundational Library",
    location="Academic Building, Floor 2",
    fee=0.00,
    study_rooms=24,
    ebook_access=True,
    max_borrow_limit=6
)

shuttle_service = ShuttleService(
    service_name="Campus Shuttle Service",
    location="Main Gate Bus Stop",
    fee=250.00,
    route_code="Moi Avenue - Main Gate - Academic Building",
    weekend_service=True
)

# Store subclass objects in one list
campus_services = [library_service, shuttle_service]

# Demonstrate polymorphism
print("Smart Campus Service System")
print("-" * 45)

for service in campus_services:
    print(service.display_service())
    print("-" * 45)
```

```python
# Demonstrate that the setter prevents negative fees

print("Testing fee validation:")
print("-" * 45)

# Update the fee using the setter method
library_service.set_fee(45.755)
print(f"Updated library service fee: kshs {library_service.get_fee():.2f}")

# Attempt to set a negative fee
try:
    library_service.set_fee(-10.00)
except ValueError as error:
    print(f"Exception caught: {error}")

# Confirm that the original fee was not changed
print(f"Fee after invalid update attempt: kshs {library_service.get_fee():.2f}")
```
